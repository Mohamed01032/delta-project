```python
import os
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.models as models
from torchvision.models import EfficientNet_B0_Weights

from sklearn.model_selection import StratifiedGroupKFold, train_test_split
from sklearn.metrics import (accuracy_score, precision_score, recall_score,f1_score, roc_auc_score, confusion_matrix,classification_report, roc_curve, fbeta_score)

import albumentations as A
from albumentations.pytorch import ToTensorV2

import cv2
from tqdm import tqdm
import seaborn as sns
from PIL import Image


def set_seed(seed=42):
    np.random.seed(seed)
    random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)


DATA_PATH = "/kaggle/input/datasets/irfansheriff/parkinsons-brain-mri-dataset/parkinsons_dataset"
assert Path(DATA_PATH).exists(), f"Data path not found: {DATA_PATH}"

paths = list(Path(DATA_PATH).glob("*/*.png"))
df = pd.DataFrame({"image_path": paths})
df["label_name"] = df["image_path"].apply(lambda p: p.parent.name)
df["label"] = df["label_name"].map({"normal": 0, "parkinson": 1}).astype(int)

def group_key(p):
    
    return p.stem.rsplit("_", 1)[0]

df["group"] = df["image_path"].apply(group_key)

print("Total images:", len(df))
print(df["label_name"].value_counts())
print("Unique groups:", df["group"].nunique())


def split_group_safe_per_class(df_all, seed=42, train_frac=0.80, test_frac=0.20):
    rng = np.random.RandomState(seed)
    parts = {}
    for lab in [0, 1]:
        sub = df_all[df_all["label"] == lab].copy()
        groups = sub["group"].drop_duplicates().tolist()
        rng.shuffle(groups)
        n = len(groups)
        n_train = max(1, int(round(train_frac * n)))
        n_test = n - n_train
        if n_test < 1:
            n_train -= 1
            n_test = 1
        g_train = set(groups[:n_train])
        g_test = set(groups[n_train:])
        parts[(lab, "train_val")] = sub[sub["group"].isin(g_train)]
        parts[(lab, "test")] = sub[sub["group"].isin(g_test)]

    train_val_df = pd.concat([parts[(0, "train_val")], parts[(1, "train_val")]], ignore_index=True)
    test_df = pd.concat([parts[(0, "test")], parts[(1, "test")]], ignore_index=True)

    train_val_df = train_val_df.sample(frac=1, random_state=seed).reset_index(drop=True)
    test_df = test_df.sample(frac=1, random_state=seed).reset_index(drop=True)
    return train_val_df, test_df

train_val_df, test_df = split_group_safe_per_class(df, seed=42, train_frac=0.80, test_frac=0.20)

print("\n=== Data splits ===")
print(f"Train+Val: n={len(train_val_df):4d} | normal={(train_val_df['label']==0).sum():3d} | parkinson={(train_val_df['label']==1).sum():3d}")
print(f"Test     : n={len(test_df):4d} | normal={(test_df['label']==0).sum():3d} | parkinson={(test_df['label']==1).sum():3d}")


IMG_SIZE = 224
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

train_transform = A.Compose([
    A.Resize(IMG_SIZE, IMG_SIZE),
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.3),
    A.RandomRotate90(p=0.3),
    A.Affine(translate_percent=0.05, scale=(0.95, 1.05), rotate=(-15, 15), p=0.5),
    A.RandomBrightnessContrast(p=0.2),
    A.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
    ToTensorV2()
])

val_transform = A.Compose([
    A.Resize(IMG_SIZE, IMG_SIZE),
    A.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
    ToTensorV2()
])

class MRIDataset(Dataset):
    def __init__(self, df_, transform=None):
        self.df = df_.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        img_path = self.df.loc[idx, "image_path"]
        label = int(self.df.loc[idx, "label"])
        try:
            img = cv2.imread(str(img_path))
            if img is None:
                raise ValueError(f"Image not found: {img_path}")
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        except Exception as e:
            print(f"Error loading {img_path}: {e}")
            img = np.zeros((IMG_SIZE, IMG_SIZE, 3), dtype=np.uint8)
        if self.transform:
            img = self.transform(image=img)["image"]
        return img, torch.tensor(label, dtype=torch.long)

def make_loaders(train_df, val_df, batch_size=16, num_workers=2):
    train_ds = MRIDataset(train_df, transform=train_transform)
    val_ds   = MRIDataset(val_df,   transform=val_transform)
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=num_workers, pin_memory=True)
    val_loader   = DataLoader(val_ds,   batch_size=batch_size, shuffle=False, num_workers=num_workers, pin_memory=True)
    return train_loader, val_loader


def get_model_with_unfreeze(num_classes=1, unfreeze_blocks=3):
    weights = EfficientNet_B0_Weights.IMAGENET1K_V1
    model = models.efficientnet_b0(weights=weights)
    
    for param in model.parameters():
        param.requires_grad = False
    
    stages = list(model.features.children())
    total_stages = len(stages)
    start_unfreeze = max(0, total_stages - unfreeze_blocks)
    for i in range(start_unfreeze, total_stages):
        for param in stages[i].parameters():
            param.requires_grad = True
   
    in_features = model.classifier[1].in_features
    model.classifier = nn.Sequential(
        nn.Dropout(p=0.5, inplace=True),
        nn.Linear(in_features, num_classes)
    )
    
    for param in model.classifier.parameters():
        param.requires_grad = True
    return model


class EarlyStopping:
    def __init__(self, patience=5, delta=0.0):
        self.patience = patience
        self.counter = 0
        self.best_score = None
        self.early_stop = False
        self.val_loss_min = float('inf')
        self.delta = delta
        self.best_model_wts = None

    def __call__(self, val_loss, model):
        score = -val_loss
        if self.best_score is None:
            self.best_score = score
            self.save_checkpoint(val_loss, model)
        elif score < self.best_score + self.delta:
            self.counter += 1
            if self.counter >= self.patience:
                self.early_stop = True
        else:
            self.best_score = score
            self.save_checkpoint(val_loss, model)
            self.counter = 0

    def save_checkpoint(self, val_loss, model):
        self.best_model_wts = model.state_dict().copy()
        self.val_loss_min = val_loss

def train_one_fold(train_df, val_df, num_epochs=30, lr=1e-4, lr_backbone=1e-5, weight_decay=1e-5):
    model = get_model_with_unfreeze(unfreeze_blocks=3).to(device)
    
    params = [
        {'params': model.classifier.parameters(), 'lr': lr},
        {'params': [p for n, p in model.named_parameters() if p.requires_grad and 'classifier' not in n], 'lr': lr_backbone}
    ]
    optimizer = optim.Adam(params, weight_decay=weight_decay)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3)
    early_stopping = EarlyStopping(patience=5)

   
    class_counts = train_df["label"].value_counts().values
    pos_weight = torch.tensor([class_counts[0] / class_counts[1]]).to(device)
    criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

    train_loader, val_loader = make_loaders(train_df, val_df, batch_size=16)

    best_val_acc = 0.0
    best_model_wts = None
    best_epoch = 0

    for epoch in range(num_epochs):
       
        model.train()
        running_loss = 0.0
        running_corrects = 0
        total = 0
        for inputs, labels in tqdm(train_loader, desc=f"Epoch {epoch+1} train", leave=False):
            inputs = inputs.to(device)
            labels = labels.float().unsqueeze(1).to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            preds = (torch.sigmoid(outputs) > 0.5).long()
            running_loss += loss.item() * inputs.size(0)
            running_corrects += torch.sum(preds == labels.long())
            total += labels.size(0)
        train_loss = running_loss / max(1, total)
        train_acc = running_corrects.double() / max(1, total)

        
        model.eval()
        val_loss = 0.0
        val_corrects = 0
        val_total = 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs = inputs.to(device)
                labels = labels.float().unsqueeze(1).to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                preds = (torch.sigmoid(outputs) > 0.5).long()
                val_loss += loss.item() * inputs.size(0)
                val_corrects += torch.sum(preds == labels.long())
                val_total += labels.size(0)
        val_loss /= max(1, val_total)
        val_acc = val_corrects.double() / max(1, val_total)

        print(f"Epoch {epoch+1:2d} | Train Loss: {train_loss:.4f} Acc: {train_acc:.4f} | Val Loss: {val_loss:.4f} Acc: {val_acc:.4f}")

        scheduler.step(val_loss)
        early_stopping(val_loss, model)

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_model_wts = model.state_dict().copy()
            best_epoch = epoch + 1

        if early_stopping.early_stop:
            print(f"Early stopping at epoch {epoch+1}. Best epoch: {best_epoch} with val acc {best_val_acc:.4f}")
            break

    if best_model_wts is not None:
        model.load_state_dict(best_model_wts)
    return model, best_val_acc, val_loader


print("\n" + "="*60)
print("Cross-validation on train_val set to find optimal threshold for high precision")
print("="*60)

n_splits = 5
sgkf = StratifiedGroupKFold(n_splits=n_splits, shuffle=True, random_state=42)

fold_val_accs = []
all_val_probs = []
all_val_labels = []

for fold, (train_idx, val_idx) in enumerate(sgkf.split(train_val_df, train_val_df["label"], groups=train_val_df["group"])):
    print(f"\nFold {fold+1}/{n_splits}")
    train_fold = train_val_df.iloc[train_idx]
    val_fold = train_val_df.iloc[val_idx]
    print(f"Train: {len(train_fold)} | Normal: {(train_fold['label']==0).sum()} | Parkinson: {(train_fold['label']==1).sum()}")
    print(f"Val:   {len(val_fold)} | Normal: {(val_fold['label']==0).sum()} | Parkinson: {(val_fold['label']==1).sum()}")

    model_fold, best_acc, val_loader_fold = train_one_fold(train_fold, val_fold, num_epochs=30)
    fold_val_accs.append(best_acc)

   
    model_fold.eval()
    probs_fold, labels_fold = [], []
    with torch.no_grad():
        for inputs, labels in val_loader_fold:
            inputs = inputs.to(device)
            outputs = model_fold(inputs)
            probs = torch.sigmoid(outputs).cpu().numpy().flatten()
            probs_fold.extend(probs)
            labels_fold.extend(labels.numpy())
    all_val_probs.extend(probs_fold)
    all_val_labels.extend(labels_fold)

fold_val_accs = [acc.cpu().item() for acc in fold_val_accs]  

print(f"\nCross-validation complete.")
print(f"Fold validation accuracies: {fold_val_accs}")
print(f"Mean validation accuracy: {np.mean(fold_val_accs):.4f} ± {np.std(fold_val_accs):.4f}")


all_val_probs = np.array(all_val_probs)
all_val_labels = np.array(all_val_labels)

thresholds = np.linspace(0.1, 0.99, 200)
best_prec = 0.0
best_thresh = 0.5
for thresh in thresholds:
    y_pred = (all_val_probs >= thresh).astype(int)
    prec = precision_score(all_val_labels, y_pred, zero_division=0)
    if prec >= 0.98:
        best_thresh = thresh
        best_prec = prec
        break
    if prec > best_prec:
        best_prec = prec
        best_thresh = thresh

print(f"Threshold selected from CV: {best_thresh:.3f} (CV Precision: {best_prec:.4f})")


print("\n" + "="*60)
print("Training final model on entire train_val set")
print("="*60)


train_final, val_final = train_test_split(
    train_val_df, test_size=0.2, stratify=train_val_df['label'], random_state=42
)
print(f"Final train: {len(train_final)} | Normal: {(train_final['label']==0).sum()} | Parkinson: {(train_final['label']==1).sum()}")
print(f"Final val:   {len(val_final)} | Normal: {(val_final['label']==0).sum()} | Parkinson: {(val_final['label']==1).sum()}")

model_final, best_val_acc, _ = train_one_fold(train_final, val_final, num_epochs=30)
print(f"Final model best validation accuracy: {best_val_acc:.4f}")


torch.save(model_final.state_dict(), "final_efficientnetb0_group_safe2.pth")
print("Model saved as final_efficientnetb0_group_safe.pth")


print("\n" + "="*60)
print("Evaluating on held-out test set")
print("="*60)

test_dataset = MRIDataset(test_df, transform=val_transform)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False, num_workers=2, pin_memory=True)

model_final.eval()
all_probs_test = []
all_labels_test = []
with torch.no_grad():
    for inputs, labels in tqdm(test_loader, desc="Testing"):
        inputs = inputs.to(device)
        outputs = model_final(inputs)
        probs = torch.sigmoid(outputs).cpu().numpy().flatten()
        all_probs_test.extend(probs)
        all_labels_test.extend(labels.numpy())

all_probs_test = np.array(all_probs_test)
all_labels_test = np.array(all_labels_test)


y_pred_test = (all_probs_test >= best_thresh).astype(int)

acc = accuracy_score(all_labels_test, y_pred_test)
prec = precision_score(all_labels_test, y_pred_test, zero_division=0)
rec = recall_score(all_labels_test, y_pred_test, zero_division=0)
f1 = f1_score(all_labels_test, y_pred_test, zero_division=0)
roc_auc = roc_auc_score(all_labels_test, all_probs_test)

print(f"\n=== Test Results (threshold={best_thresh:.3f}) ===")
print(f"Accuracy : {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f}")
print(f"F1-score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")
print("\nClassification Report:")
print(classification_report(all_labels_test, y_pred_test, target_names=["Normal", "Parkinson"]))

cm = confusion_matrix(all_labels_test, y_pred_test)
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=["Normal", "Parkinson"], yticklabels=["Normal", "Parkinson"])
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title(f"Confusion Matrix - Threshold={best_thresh:.2f}")
plt.show()


fpr, tpr, _ = roc_curve(all_labels_test, all_probs_test)
plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f"AUC = {roc_auc:.2f}")
plt.plot([0,1], [0,1], 'k--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()

print("\nAll done.")
```

```python
import torch
import torch.nn as nn
import torchvision.models as models
from torchvision.models import EfficientNet_B0_Weights
import cv2
import numpy as np
import albumentations as A
from albumentations.pytorch import ToTensorV2
from pathlib import Path


def get_model_with_unfreeze(num_classes=1, unfreeze_blocks=3):
    weights = EfficientNet_B0_Weights.IMAGENET1K_V1
    model = models.efficientnet_b0(weights=weights)
   
    for param in model.parameters():
        param.requires_grad = False
    
    stages = list(model.features.children())
    total_stages = len(stages)
    start_unfreeze = max(0, total_stages - unfreeze_blocks)
    for i in range(start_unfreeze, total_stages):
        for param in stages[i].parameters():
            param.requires_grad = True
   
    in_features = model.classifier[1].in_features
    model.classifier = nn.Sequential(
        nn.Dropout(p=0.5, inplace=True),
        nn.Linear(in_features, num_classes)
    )
    return model


IMG_SIZE = 224
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

val_transform = A.Compose([
    A.Resize(IMG_SIZE, IMG_SIZE),
    A.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
    ToTensorV2()
])


def predict_image(model, image_path, threshold=0.927, device='cuda'):
    """
    Loads and preprocesses a single image, runs inference, and returns prediction.
    
    Args:
        model: the trained PyTorch model (already loaded)
        image_path: path to the image file (PNG)
        threshold: decision threshold for Parkinson class (default 0.927)
        device: 'cuda' or 'cpu'
    
    Returns:
        dict with keys:
            'class': 'Normal' or 'Parkinson'
            'probability': probability of being Parkinson (float)
            'confidence': max probability (float)
    """
    
    img = cv2.imread(str(image_path))
    if img is None:
        raise ValueError(f"Could not load image: {image_path}")
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    
    transformed = val_transform(image=img)
    input_tensor = transformed["image"].unsqueeze(0).to(device)  # add batch dimension
    
    
    model.eval()
    with torch.no_grad():
        output = model(input_tensor)
        prob = torch.sigmoid(output).cpu().item()  # probability of Parkinson
    
    
    pred_class = "Parkinson" if prob >= threshold else "Normal"
    
    return {
        'class': pred_class,
        'probability': prob,
        'confidence': max(prob, 1 - prob)  # highest probability among the two classes
    }


if __name__ == "__main__":
    # Set device
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print("Using device:", device)
    
    # Load the trained model
    model_path = "/kaggle/working/final_efficientnetb0_group_safe2.pth"  # adjust path if needed
    model = get_model_with_unfreeze(num_classes=1, unfreeze_blocks=3)
    model.load_state_dict(torch.load(model_path, map_location=device))
    model.to(device)
    print("Model loaded successfully.")
    
    test_image = "/kaggle/input/datasets/irfansheriff/parkinsons-brain-mri-dataset/parkinsons_dataset/normal/Mag_Images_002.png"
    result = predict_image(model, test_image, threshold=0.927, device=device)
    print(result)
    
```
